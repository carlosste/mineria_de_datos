## Task 6: Grouping

Use groupby to aggregate data by category.

```python
# 6.1 Group by category and calculate total value per category
# Hint: df.groupby('CATEGORY_COLUMN')['VALUE_COLUMN'].sum()
# YOUR CODE HERE
category_totals = df.groupby("CUENTA")["Valor (Miles)"].sum()
```

```python
# 6.2 Sort the results from highest to lowest
# Hint: .sort_values(ascending=False)
# YOUR CODE HERE
category_totals = category_totals.sort_values(ascending=False)
```

```python
# 6.3 Display the top 5 categories by total value
# Save the result to a variable for the visualization
# YOUR CODE HERE
top_5_categories = category_totals.head(5)

top_5_categories
# top_5_categories = ...
```

### Task 6 Observations

- What are the top 5 categories by total value?
- Is there a category that dominates the total value?
- How does the distribution look? (concentrated or spread out?)

Se puede observar que algunas cuentas concentran gran parte de los ingresos, lo que muestra una distribución relativamente concentrada.

## Task 7: Visualization

Create a bar chart showing the top 10 categories by total value.

```python
# 7.1 Create a bar chart of top 10 categories by total value
#
# Steps:
# 1. Group by category, sum values, sort descending, take top 10
# 2. Use .plot(kind='bar') to create the chart
# 3. Add title and labels
#
# Example:
# top_10 = df.groupby('CATEGORY')['VALUE'].sum().sort_values(ascending=False).head(10)
# top_10.plot(kind='bar', figsize=(12, 6), color='steelblue')
# plt.title('Top 10 Categories by Total Value')
# plt.xlabel('Category')
# plt.ylabel('Total Value')
# plt.xticks(rotation=45, ha='right')
# plt.tight_layout()
# plt.show()

# YOUR CODE HERE
top_10 = df.groupby("CUENTA")["Valor (Miles)"] \
           .sum() \
           .sort_values(ascending=False) \
           .head(10)

top_10.plot(kind="bar", figsize=(12,6))

plt.title("Top 10 Categories by Total Value")
plt.xlabel("Category")
plt.ylabel("Total Value (Thousands)")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

### Task 7 Observations

- What does the visualization tell you about the data distribution?
- Which categories stand out?

La visualización muestra que algunas categorías tienen valores significativamente más altos que otras.
Las categorías principales destacan claramente, lo que indica que una parte importante del valor total se concentra en pocas cuentas.
